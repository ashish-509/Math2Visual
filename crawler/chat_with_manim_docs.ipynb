```python
with open("markdown_output.md", "r") as f:
    docs_text = f.read()
```

```python
from google import genai
import os
from dotenv import load_dotenv

load_dotenv()
client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))
```

```python
response = client.models.generate_content(
    model = "gemini-2.5-flash",
    contents = f""" You are a expert manim code generator, who generates the manim code on the basis of
      prompt provided. The reference for you is the documentation of the mainm community -> {docs_text}. 
       
        draw two tangents from a point outside a circle and show they're equal
          """
      )
print(response.text)
```
